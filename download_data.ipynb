```python
import requests
import pandas as pd
import os
from bs4 import BeautifulSoup
```

```python


for i in range (2024, 2025):
    for j in range(1, 13):
        print(i)
        print(j)
        if i<2020:
            link_rrc = 'https://www.rrc.texas.gov/oil-and-gas/research-and-statistics/refinery-statements/refineries-statements-{i}/refinery-statements-{j:02d}-{i}'.format(i=i, j=j)
        else:
            link_rrc = 'https://www.rrc.texas.gov/oil-and-gas/research-and-statistics/refinery-statements/refineries-statements-{i}/refinery-statements-{j}-{i}'.format(i=i, j=j)
        response = requests.get(link_rrc)
        soup = BeautifulSoup(response.text, 'html.parser')
        parsed_table = soup.find_all('table', {'class': 'table table-bordered'})[0]

        data = []
        for row in parsed_table.find_all('tr'):
            row_data = []
            for td in row.find_all('td'):
                text_filed = td.text.strip()
                row_data.append(text_filed)
                td_check = td.find('a')
                if td_check is not None:
                    link = td.a['href']
                    row_data.append(link)
                else:
                    not_link = 'No link'
                    row_data.append(not_link)
            data.append(row_data)
        df = pd.DataFrame(data[1:], columns=['facilityid', 'link1', 'operator', 'link2'])
        df['link'] = df['link1']
        df.loc[df['link1']=='No link', 'link'] = df['link2']
        df = df[(df['link'] != 'No link')].dropna(how='all')
        df.loc[~(df['link'].str.contains('http')), 'link'] = 'https://www.rrc.texas.gov' + df['link']

        for index, row in df.iterrows():
            link = row['link']
            facilityid = row['facilityid']
            operator = row['operator']

            response = requests.get(link)

            os.makedirs('Raw Data/Texas RRC/{}'.format(facilityid), exist_ok = True)

            pdf = open('Raw Data/Texas RRC/{facilityid}/{i}_{j}___{facilityid}_{operator}.pdf'.format(i=i, j=j, facilityid=facilityid, operator=operator), 'wb')
            pdf.write(response.content)
            pdf.close()
 
```
